##**Tratamiento de los Datos**

```python
#Eliminando Columnas que no aportan Valor

df_datos.drop(['customerID','account_PaperlessBilling','account_PaymentMethod','valor'],axis=1,inplace=True)
df_datos.head()
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
#Encoding - Transformando variables categóricas a formato númerico

df_datos['Churn'] = df_datos['Churn'].replace({'No':0,'Yes':1})
df_datos['customer_gender'] = df_datos['customer_gender'].replace({'Male':0,'Female':1})
df_datos['customer_Partner'] = df_datos['customer_Partner'].replace({'No':0,'Yes':1})
df_datos['customer_Dependents'] = df_datos['customer_Dependents'].replace({'No':0,'Yes':1})
df_datos
```

```python
#Encoding - Transformando variables categóricas a formato númerico

categoricas = ['account_Contract','tipo_de_servicio']
datos_codif = pd.get_dummies(data=df_datos,columns=categoricas,dtype=int)
datos_codif
```

```python
#Proporción de Cancelación (Churn)

Dist_Churn = datos_codif.Churn.value_counts()
Dist_Churn
```

```python
import numpy as np

def ancho_bin(df,columna):
  q75,q25 = np.percentile(df[columna],[75,25])
  iqr = q75-q25
  ancho = 2 * iqr * np.power(len(df[columna]), -1/3)
  return ancho
```

```python
cargo_mensual_promedio = df_datos['account_Charges.Monthly'].mean()
cargo_mensual_mediana = df_datos['account_Charges.Monthly'].median()

fig, ax = plt.subplots(1,2,figsize=(9,4))

sns.boxplot(data=df_datos, y='account_Charges.Monthly',ax=ax[0])
ax[0].set_title('Boxplot')
ax[0].axhline(y=cargo_mensual_promedio, color='r', linestyle='--', label='Promedio')
ax[0].legend()

binwidth = ancho_bin(df_datos, 'account_Charges.Monthly')
sns.histplot(data=df_datos, x='account_Charges.Monthly', ax=ax[1], kde=True, binwidth=binwidth)
plt.ylabel('Número de vuelos')
plt.grid(False)
ax[1].set_title('Histograma')
ax[1].axvline(x=cargo_mensual_promedio, color='r', linestyle='--', label='Promedio')
ax[1].axvline(x=cargo_mensual_mediana, color='y', linestyle='--', label='Mediana')
ax[1].legend()

plt.tight_layout()

plt.show()
```

##**Separación de datos para Modelos Predictivos**

###**Variables explicativas y variable de respuesta**

```python
X=datos_codif.drop('Churn',axis=1)
y=datos_codif.Churn
```

###**Ajustando el modelo y Creación de Conjunto de Datos**

Conjuntos de Prueba, Entrenamiento y Validación

```python
from sklearn.model_selection import train_test_split

#X_train, X_test, y_train, y_test = train_test_split(X, y, stratify = y,  random_state = 5)

X,X_test,y,y_test = train_test_split(X,y,test_size=0.15,stratify=y,random_state=5)
X_train,X_val,y_train,y_val = train_test_split(X,y,stratify=y,random_state=5)

```

##**Creación de Modelos Predictivos**

Crea al menos dos modelos diferentes para predecir la cancelación de clientes.

Un modelo puede requerir normalización, como Regresión Logística o KNN.

El otro modelo puede no requerir normalización, como Árbol de Decisión o Random Forest.

La decisión de aplicar o no la normalización depende de los modelos seleccionados. Ambos modelos pueden ser creados sin normalización, pero también es una opción combinar modelos con y sin normalización.

*Justificación:*

Regresión Logística / KNN: Estos modelos son sensibles a la escala de los datos, por lo que la normalización es importante para que los coeficientes o las distancias se calculen correctamente.

Árbol de Decisión / Random Forest: Estos modelos no dependen de la escala de los datos, por lo que no es necesario aplicar normalización.

###**Modelo de referencia - Baseline**



Este algoritmo servirá como un criterio de comparación para identificar si los otros modelos tienen un rendimiento mejor que la clasificación más simple posible.

```python
from sklearn.dummy import DummyClassifier

dummy = DummyClassifier()
dummy.fit(X_train, y_train)

dummy.score(X_test, y_test)
```

###**Modelo 1 - Desicion Tree**

El modelo de árbol de decisión es muy utilizado debido a su alta **explicabilidad** y **procesamiento rápido**, manteniendo un rendimiento bastante interesante.

Se basa en decisiones simples tomadas por el algoritmo, separando los datos mediante comparaciones de **menor** y **mayor** en los valores de las columnas de la base de datos.

```python
from sklearn.tree import DecisionTreeClassifier

modelo_arbol = DecisionTreeClassifier(max_depth = 5, random_state =5)
modelo_arbol.fit(X_train, y_train)

modelo_arbol.score(X_val, y_val)

#Validando el modelo

print(f'La exactitud del modelo_arbol con el conjunto de entrenamiento es de: {modelo_arbol.score(X_train,y_train)}')
print(f'La exactitud del modelo_arbol con el conjunto de validación es de: {modelo_arbol.score(X_val,y_val)}')
```

###**Normalización**

Evalúa la necesidad de normalizar o estandarizar los datos, según los modelos que se aplicarán. Modelos basados en distancia, como KNN, SVM, Regresión Logística y Redes Neuronales, requieren este preprocesamiento. Por otro lado, modelos basados en árboles, como Decision Tree, Random Forest y XGBoost, no son sensibles a la escala de los datos.

```python
#Datos de entrenamiento

from sklearn.preprocessing import MinMaxScaler

normalizacion = MinMaxScaler()
X_train_normalizado = normalizacion.fit_transform(X_train)

pd.DataFrame(X_train_normalizado)
```

###**Modelo 2 - KNN**

El algoritmo KNN se basa en el cálculo de la distancia entre los registros de la base de datos y busca elementos que estén cerca unos de otros (vecinos) para tomar la decisión de clasificación.

Debido a que utiliza cálculos de distancia, este algoritmo está influenciado por la escala de las variables, y por eso es necesario realizar una transformación en los datos antes de utilizar este método.

```python
from sklearn.neighbors import KNeighborsClassifier

modelo_knn = KNeighborsClassifier(n_neighbors=5)

modelo_knn.fit(X_train_normalizado, y_train)

X_val_normalizado = normalizacion.transform(X_val)
X_test_normalizado = normalizacion.transform(X_test)

y_val_previsto2 = modelo_knn.predict(X_val_normalizado)

#modelo_knn.score(X_test_normalizado, y_test)

#Validando el modelo

print(f'La exactitud del modelo_knn con el conjunto de entrenamiento es de: {modelo_knn.score(X_train_normalizado,y_train)}')
print(f'La exactitud del modelo_knn con el conjunto de validación es de: {modelo_knn.score(X_val_normalizado,y_val)}')

```

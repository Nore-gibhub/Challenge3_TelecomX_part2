##**Tratamiento de los Datos**

```python
#Eliminando Columnas que no aportan Valor

df_datos.drop(['customerID','account_PaperlessBilling','account_PaymentMethod','valor'],axis=1,inplace=True)
df_datos.head()
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
#Encoding - Transformando variables categóricas a formato númerico

df_datos['Churn'] = df_datos['Churn'].replace({'No':0,'Yes':1})
df_datos['customer_gender'] = df_datos['customer_gender'].replace({'Male':0,'Female':1})
df_datos['customer_Partner'] = df_datos['customer_Partner'].replace({'No':0,'Yes':1})
df_datos['customer_Dependents'] = df_datos['customer_Dependents'].replace({'No':0,'Yes':1})
df_datos
```

```python
#Encoding - Transformando variables categóricas a formato númerico

categoricas = ['account_Contract','tipo_de_servicio']
datos_codif = pd.get_dummies(data=df_datos,columns=categoricas,dtype=int)
datos_codif
```

```python
X=datos_codif.drop('Churn',axis=1)
y=datos_codif.Churn
```

```python
#Proporción de Cancelación (Churn)

Dist_Churn = datos_codif.Churn.value_counts()
Dist_Churn
```

```python
#Proporción de Cancelación en % (Churn)

Dist_Churn_Porcentaje = (datos_codif.Churn.value_counts(normalize=True).round(3)*100).map("{:.1f}%".format)
Dist_Churn_Porcentaje
```

```python
from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier()
modelo.fit(X,y)
modelo.score(X,y)

print(f'La exactitud del modelo es: {modelo.score(X,y)}')
```

##**Balanceo de Clases**

```python
from imblearn.over_sampling import SMOTE

over_sampling = SMOTE()
x_balanceada, y_balanceada = over_sampling.fit_resample(X,y)

y_balanceada.value_counts(normalize=True)
```

```python
import numpy as np

def ancho_bin(df,columna):
  q75,q25 = np.percentile(df[columna],[75,25])
  iqr = q75-q25
  ancho = 2 * iqr * np.power(len(df[columna]), -1/3)
  return ancho
```

```python
cargo_mensual_promedio = df_datos['account_Charges.Monthly'].mean()
cargo_mensual_mediana = df_datos['account_Charges.Monthly'].median()

fig, ax = plt.subplots(1,2,figsize=(9,4))

sns.boxplot(data=df_datos, y='account_Charges.Monthly',ax=ax[0])
ax[0].set_title('Boxplot')
ax[0].axhline(y=cargo_mensual_promedio, color='r', linestyle='--', label='Promedio')
ax[0].legend()

binwidth = ancho_bin(df_datos, 'account_Charges.Monthly')
sns.histplot(data=df_datos, x='account_Charges.Monthly', ax=ax[1], kde=True, binwidth=binwidth)
plt.ylabel('Número de vuelos')
plt.grid(False)
ax[1].set_title('Histograma')
ax[1].axvline(x=cargo_mensual_promedio, color='r', linestyle='--', label='Promedio')
ax[1].axvline(x=cargo_mensual_mediana, color='y', linestyle='--', label='Mediana')
ax[1].legend()

plt.tight_layout()

plt.show()
```

Contrato a Dos Años (-0.30): Es una de las anclas más fuertes. El compromiso legal y la estabilidad del servicio retienen al cliente.

Antigüedad (-0.35): Es una correlación negativa potente. Significa que mientras más meses lleva un cliente en la compañía, menor es la probabilidad de que se vaya. Los clientes nuevos son los más volátiles; una vez que pasan la barrera de los primeros meses, se vuelven leales.

3. El factor económico
Cargos Mensuales (0.19): A medida que sube la factura, sube el Churn. Es la típica sensibilidad al precio.

Cargos Totales (-0.19): Esto parece contradictorio pero no lo es. Los cargos totales altos suelen estar amarrados a una antigüedad alta. Por eso, aunque el total pagado sea mucho, ese cliente es un "sobreviviente" que ya demostró ser fiel.
